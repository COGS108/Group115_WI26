## Project Proposal
RESEARCH QUESTION
How has the increase in AI-related demand for High Bandwidth Memory (HBM) since 2022 affected the price per gigabyte of consumer DDR4 and DDR5 RAM in the United States between 2018 and 2025?

Variables:
Independent: AI demand (proxied by HBM production or NVDA stock)

Dependent: RAM price per GB

## BACKGROUND & PRIOR WORK:
The semiconductor memory market historically follows cyclical "boom and bust" patterns driven by supply and demand. Typically, as manufacturing technology matures, yields improve, and prices for consumer RAM (DDR4/DDR5) decrease over time. However, the unprecedented explosion of Generative AI since late 2022 has introduced a new variable. Major manufacturers like Samsung and SK Hynix have reportedly shifted significant production capacity away from standard consumer DRAM to focus on High Bandwidth Memory (HBM) required for AI GPUs, potentially creating a "supply shock" for the consumer market.
Prior work in this domain often focuses on general hardware price trends, such as the impact of cryptocurrency mining on GPU prices (a demand-side shock). There is less existing analysis on how the AI server boom creates a "crowding out" effect for the average PC user's memory components. Our project aims to fill this gap by analyzing whether the AI boom has disrupted the expected historical price per gigabyte over time of computer memory.
Prior work on GPU pricing during the cryptocurrency mining boom provides a useful comparison for understanding demand-driven price shocks in hardware markets. For example, increases in cryptocurrency prices led to higher demand for GPUs, which in turn raised consumer prices. Similarly, the current AI boom may act as a supply-side shock, leading manufacturing to shift toward HBM and reducing the availability of consumer RAM. By comparing these two cases, we can better understand whether AI-driven changes follow similar or distinct pricing patterns.

## 4. Hypothesis
We hypothesize that the price per gigabyte of consumer RAM decreases over time following the release of a new generation (DDR4/DDR5), but that this decline slows or temporarily reverses after 2022 due to increased demand for HBM driven by AI applications.

We will test this by comparing the slope of price decline before and after 2022. If the rate of price decrease becomes significantly smaller or positive after 2022, we will reject the standard depreciation model and conclude that AI demand has disrupted historical pricing trends.

Reasoning: New manufacturing processes (like DDR5) initially have low yields and high R&D recovery costs. As production scales and yields improve, prices drop rapidly before stabilizing. We expect high-speed, low-latency kits to retain value longer than standard "budget" kits.

## 5. Data
We will use a historical RAM pricing dataset sourced from PCPartPicker, available through Kaggle archives. This dataset contains time-series price data for consumer RAM kits collected from major U.S. retailers such as Amazon and Newegg.

The dataset includes the following variables:
- Date (weekly or daily observations)
- Price (USD)
- Capacity (GB)
- RAM Type (DDR4, DDR5)
- Speed (MHz)
- CAS Latency (CL timing)

We will transform the data to compute price per gigabyte (price/GB) to standardize comparisons across different products.

Our analysis will focus on the time period from 2018 to 2025 to capture both pre-AI (before 2022) and post-AI expansion trends. The dataset contains over 1000 observations, which is sufficient for time-series analysis and regression modeling.
